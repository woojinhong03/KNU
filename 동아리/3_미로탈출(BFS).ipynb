```python
n, m = map(int, input().split())

miro = [list(map(int, input())) for _ in range(n)]

px = [-1, 1, 0, 0]
py = [0, 0, -1, 1]

queue = [(0, 0)]

while queue:
    x, y = queue.pop(0)

    for i in range(4):
        new_x = x + px[i]
        new_y = y + py[i]

        if new_x < 0 or new_y < 0 or new_x >= n or new_y >= m or miro[new_x][new_y] == 0:
            continue

        if miro[new_x][new_y] == 1:
            miro[new_x][new_y] = miro[x][y] + 1
            queue.append((new_x, new_y))

print(miro[n - 1][m - 1])
```
