```python
n, m = 4, 6
dduck = [19, 15, 10, 17]

start = 0
end = max(dduck)

while(start <= end):
    mid = (start + end) // 2
    tot = sum(i - mid for i in dduck if mid < i)

    if tot < m:
        end = mid - 1

    else:
        dap = mid
        start = mid + 1

print(dap)
```
