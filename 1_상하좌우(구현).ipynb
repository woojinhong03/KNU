```python
# map_size = 5
# move = ['R', 'R', 'R', 'U', 'D', 'D']
map_size = int(input())
move = input().split()

start = [1, 1]

pos = {
    'L' : [0, -1],
    'R' : [0, 1],
    'U' : [-1, 0],
    'D' : [1, 0]
}

for i in move:
    next_pos = [start[0] + pos[i][0], start[1] + pos[i][1]]

    if next_pos[0] < 1 or next_pos[0] > map_size or next_pos[1] < 1 or next_pos[1] > map_size:
        continue

    start = next_pos

print(start)
```
