```python
n, m = map(int, input().split())

distance = [[int(1e9)] * (n+1) for _ in range(n + 1)]

for i in range(1, n+1):
    distance[i][i] = 0

for _ in range(m):
    a, b = map(int, input().split())
    distance[a][b], distance[b][a] = 1, 1

x, k = map(int, input().split())

for mid in range(1, n+1):
    for start in range(1, n+1):
        for end in range(1, n+1):
            distance[start][end] = min(distance[start][end], distance[start][mid] + distance[mid][end])

if distance[1][k] + distance[k][x] >= int(1e9):
    print(-1)
else:
    print(distance[1][k] + distance[k][x])
```
